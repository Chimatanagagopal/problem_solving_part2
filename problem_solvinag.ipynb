#Problem 6 — Largest difference

Problem 7 — Move all zeros to the end

```python
numbers = [0, 1, 0, 3, 12, 0, 5]
nums=0
zeros=0
for i in numbers:
  if i!=0:
    numbers[nums]=i
    nums+=1
  else:
    zeros+=1

for i in range(zeros):
  numbers[nums]=0
  nums+=1
print(numbers)
```

Problem 9 — Find the missing number

Problem 14 — Find the longest consecutive sequence

```python
numbers = [100, 4, 200, 1, 3, 2]

longest = 0

for i in numbers:
    count = 1
    current = i

    while current + 1 in numbers:
        current += 1
        count += 1

    if count > longest:
        longest = count

print(longest)
```

Problem 16 — Check if two strings are anagrams

```python
text1 = "listen"
text2 = "silent"
dict1={}
found=False
if (len(text1)!=len(text2)):
  print("not anagarams")
else:
  for i in text1:
    if i not in dict1:
      dict1[i]=1
    else:
      dict1[i]+=1

  for j in text2:
    if j in dict1:
      dict1[j]-=1
    else:
      found=True
      break
  for k,v in dict1.items():
    if v!=0:
      found=True
      break
  print("it's not anagram" if found else "it's an anagram")
```
