Problem 16 — Check if two strings are anagrams

```python
text1 = "listen"
text2 = "silent"
dict1={}
found=False
if (len(text1)!=len(text2)):
  print("not anagarams")
else:
  for i in text1:
    if i not in dict1:
      dict1[i]=1
    else:
      dict1[i]+=1

  for j in text2:
    if j in dict1:
      dict1[j]-=1
    else:
      found=True
      break
  for k,v in dict1.items():
    if v!=0:
      found=True
      break
  print("it's not anagram" if found else "it's an anagram")
```
